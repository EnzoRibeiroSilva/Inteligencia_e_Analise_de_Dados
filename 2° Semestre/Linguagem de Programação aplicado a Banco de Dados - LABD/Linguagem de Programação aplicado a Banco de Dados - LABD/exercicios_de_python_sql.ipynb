## **Parte 4: Transações e ROLLBACK**

- Explicação rápida: As transações possuem 3 conceitos
    - BEGIN -> Início da transação
    - COMMIT -> Confirmação de que as alterações ocorrem
    - ROLLBACK -> Desfaz alterações apenas em casos em que erros acontecem

Então, vamos simular um erro que ocorreu na inserção de um novo funcionários, onde seu cargo estará vazio!

```python
try:
    conn.execute("BEGIN;")
    cursor.execute("INSERT INTO funcionarios (nome, cargo, salario, data_contratacao) VALUES('Kevin', 'Estagiário', 4200, '25/09/2024');")
    raise Exception("Erro Simulado!")
    conn.commit()
except:
    conn.rollback()
    print("Funcionário Não Inserido!")
```

- Aqui temos um erro orgânico, utilizando commit

```python
try:
    conn.execute("BEGIN;")
    cursor.execute("INSERT INTO funcionarios (nome, cargo, salario, data_contratacao) VALUES('Kevin', 4200, '25/09/2024');")
    conn.commit()
except:
    conn.rollback()
    print("Funcionário não inserido! - Transação revertida por erro!")
```

- Verificação de segurança da tabela

```python
cursor.execute("SELECT * FROM funcionarios;")
cursor.fetchall()
```

```python
cursor.execute("""
    SELECT f.nome, p.nome_projeto FROM funcionarios f
    LEFT JOIN funcionarios_projetos f_p
    ON f.id = f_p.funcionario_id
    LEFT JOIN projetos p
    ON f_p.projeto_id = p.id
""")
cursor.fetchall()
```

## **Parte 6: DCL (Controle de Dados)**

Diferença entre: DDL, DML, DQL e DCL:

- DDL: É responsável pela criação, alteração e exclusão de tabelas, portanto, é a parte resposável por cuidar da estrutura do banco. Exemplo: CREATE, ALTER e DROP;

- DML: É responsável por cuidar dos dados dentro das tabelas. Então é a área que insere, altera ou remove registros. Exemplo: INSERT, UPDATE e DELETE;

- DQL: Serve para consultar os dados. Exemplo: SELECT;

- DCL: Cuida da segurança e do controle de acesso. Exemplo: GRANT e REVOKE.

Portanto, o DCL é extremamente necessário para controle de hierarquia e acesso aos dados no banco de dados das empresas, é extremamente relevante para reforçar a segurança dele!
