Exercício 13

```python
clientes = []
for i in range(0,4):
    quant_client = int(input("Digite quantos clientes atendidos:"))
    clientes.append(quant_client)

total = sum(clientes)
media = total / 4

print(f"Média diária de atendimentos: {media}\nTotal de atendimentos no período de 4 dias: {total}")
```

Exercício 14

```python
faturamento = []
for i in range(5):
    vend_fat = float(input(f"Digite o faturamento do vendedor {i+1}:"))
    faturamento.append(vend_fat)

maximo = max(faturamento)
print(f"Faturamento máximo entre os vendedores: {maximo}\nFaturamento médio por vendedor: {sum(faturamento) / 5}")
```
