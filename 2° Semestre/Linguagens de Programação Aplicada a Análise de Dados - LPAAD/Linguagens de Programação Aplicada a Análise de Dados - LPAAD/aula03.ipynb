Texto Colorido

```python
def texto_colorido(texto, cor):
    if cor == "vermelho":
        teste = 31

    elif cor == "verde":
        teste = 32

    elif cor == "amarelo":
        teste =  33

    elif cor == "azul":
        teste = 34

    elif cor == "magenta":
        teste = 35

    elif cor == "ciano":
        teste = 36

    padrao = f"\033[{teste}m{texto}\33[0m"
    return padrao
```

```python
mensagem = "Cuidado!"

print(texto_colorido(mensagem, "vermelho"))
```

Biblioteca de Pausa/Tempo

```python
import time
```

```python
def carregar_sistema():
    print(f"{texto_colorido("Iniciando as operações do sistema...", "azul")}")
    time.sleep(5)

    print(f"{texto_colorido("Carregando pacotes pendentes...", "amarelo")}")
    time.sleep(5)

    print(f"{texto_colorido("Carregamento finalizado...", "verde")}")
    time.sleep(5)

    print("Aproveite o sistema!")
```

```python
carregar_sistema()
```

Cálculo de Mediana

```python
conjunto_a = [23,14,31,17,20]
conjunto_b = [75,1,4,3,19,50,64,31]

def calcular_mediana(lista):
    lista_ordenada = sorted(lista)
    quant = len(lista_ordenada)
    meio = (quant // 2)

    if quant % 2 == 0:
        meio -= 1
        mediana = (lista_ordenada[meio] + lista_ordenada[meio + 1]) / 2
        return mediana
    
    else:
        mediana = lista_ordenada[meio]
        return mediana
```

```python
calcular_mediana(conjunto_a)
```

```python
calcular_mediana(conjunto_b)
```

```python
calcular_mediana([1,2,3,4,5,6,7,8,9])
```

Cálculo de Moda

```python
def calcular_moda(lista):
    frequencia = {}
    for i in lista:
        frequencia[i] = lista.count(i)
    for chave, valor in frequencia.items():
        print(f"Número: {chave} - Quantidade encontrada: {valor}")     
```

```python
valores = [12,15,18,15,20,15,17]
calcular_moda(valores)
```
