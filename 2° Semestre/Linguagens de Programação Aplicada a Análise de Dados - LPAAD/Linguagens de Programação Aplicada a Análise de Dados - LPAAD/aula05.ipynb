## **Prática: Implemente calcula_variancia()**

```python
def calcula_media(lista):
    soma = sum(lista)
    media = soma / len(lista)
    return media
```

```python
conjunto = [12, 18, 15, 21, 14, 16]
```

```python
def calcula_variancia(lista):
    media = calcula_media(lista)
    cont = 0

    for i in range(len(lista)):
        var = pow((lista[i] - media), 2)
        cont += var

    resultado = cont / len(lista)
    return resultado
```

```python
calcula_variancia(conjunto)
```

## **Prática: implemente calcula_quartis() - Simples**

```python
def calcula_quartis(lista):
    lista = sorted(lista)
    n = len(lista)
    posicoes = [
        (n + 1) / 4,
        (n + 1) / 2,
        3 * (n + 1) / 4
    ]

    quartis = []

    for posicao in posicoes:
        pos_int = int(posicao)
        if posicao == pos_int:
            v_quartil = lista[pos_int - 1]

        else:
            # decimal = posicao - pos_int
            p_ant = lista[pos_int - 1]
            p_suc = lista[pos_int]

            # diferenca = p_suc - p_ant
            # proporcao = (p_suc - p_ant) * (posicao - pos_int)
            v_quartil = p_ant + ((p_suc - p_ant) * (posicao - pos_int))

        quartis.append(v_quartil)

    return quartis
    
```

```python
vendas = [42,18,35,27,50,31,22]
```

```python
q1, q2, q3 = calcula_quartis(vendas)
```

```python
print(f"Q1: {q1}")
print(f"Q2: {q2}")
print(f"Q3: {q3}")
```

## **Prática: Intervalo Interquartil (IQR)**

```python
def calcular_iqr(lista):
    q1, q2, q3 = calcula_quartis(lista)
    iqr = q3 - q1
    return iqr
```

```python
calcular_iqr(vendas)
```
