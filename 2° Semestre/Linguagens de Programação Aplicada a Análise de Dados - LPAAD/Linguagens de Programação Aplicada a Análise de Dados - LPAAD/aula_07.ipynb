## **Problema 1: Qual equipe teve melhor desempenho?**

```python
import statistics as st
```

```python
maquina_a = [98,101,100,99,102,100,101]
maquina_b = [82,115,91,108,120,86,99]
```

```python
def calcula_maquinas(lista):
    media = st.mean(lista)
    variancia = st.pvariance(lista)
    desvio = st.pstdev(lista)
    amplitude = max(lista) - min(lista)
    c_var = (desvio / media) * 100
    return media, variancia, desvio, amplitude, c_var
```

```python
media, variancia, desvio, amplitude, c_var = calcula_maquinas(maquina_a)

print(f"Média: {media:.2f}")
print(f"Variância: {variancia:.2f}")
print(f"Desvio: {desvio:.2f}")
print(f"Amplitude: {amplitude}")
print(f"Coeficiente de variação: {c_var:.2f}%")
```

```python
media, variancia, desvio, amplitude, c_var = calcula_maquinas(maquina_b)

print(f"Média: {media:.2f}")
print(f"Variância: {variancia:.2f}")
print(f"Desvio: {desvio:.2f}")
print(f"Amplitude: {amplitude}")
print(f"Coeficiente de variação: {c_var:.2f}%")
```

**Discussão**: Observando os resultados obtidos acima, (Sem considerar o Coeficiente de variação), eu percebo que a maquina_b possui a mesma média que a maquina_a, porém a maquina_b possui uma amplitude muito maior (38), assim como o seu desvio padrão (13.58). Portanto, é possível notar que o intervalo dos dados é bem descrepante de um para o outro.

- O coeficiente de variação da segunda máquina está acima de 13%, portanto a variabilidade é muito alta, pensando que a primeira máquina possui o coeficiente de variabilidade de aproximadamente 1%, demonstrando a homogeneidade dos dados.

## **Problema 2: Qual máquina é mais regular?**


```python
tempos = [48, 32, 55, 41, 37, 62, 29, 45, 51, 35, 70]
```

```python
def calcula_quartis(lista):
    dados_ordenados = sorted(lista)
    n = len(lista)
    posicoes = [
        (n + 1) / 4,
        (n + 1) / 2,
        (3 * (n + 1)) / 4
    ]

    quartis = []

    for posicao in posicoes:
        posicao_inteira = int(posicao)
        if posicao == posicao_inteira:
            valor_quartil = dados_ordenados[posicao_inteira - 1]
        else:
            parte_decimal = posicao - posicao_inteira
            valor_inferior = dados_ordenados[posicao_inteira - 1]
            valor_superior = dados_ordenados[posicao_inteira]

            diferenca = valor_superior - valor_inferior
            proporcao = diferenca * parte_decimal

            valor_quartil = valor_inferior + proporcao
        quartis.append(valor_quartil)
    return quartis
```

```python
def calcula_tempo(lista):
    media = st.mean(lista)
    mediana = st.median(lista)
    amplitude = max(lista) - min(lista)
    variancia = st.pvariance(lista)
    desvio_padrao = st.pstdev(lista)
    return media, mediana, amplitude, variancia, desvio_padrao
```

```python
media, mediana, amplitude, variancia, desvio = calcula_tempo(tempos)
q1, q2, q3 = calcula_quartis(tempos)
iqr = q3 - q1
```

```python
print(f"Média: {media:.2f}")
print(f"Mediana: {mediana}")
print(f"Amplitude: {amplitude}")
print(f"Variancia: {variancia:.2f}")
print(f"Desvio Padrão: {desvio:.2f}")
print(f"q1: {q1}")
print(f"q2: {q2}")
print(f"q3: {q3}")
print(f"iqr: {iqr}")
```

## **Prática - Situação**: Uma empresa registrou datas no sistema como string, e nós precisamos fazer cálculos e demonstrar as datas aqui!

```python
data_pedido = "12/09/2026"
data_entrega = "18/09/2026"
```

```python
def conversao_data(data):
    data = datetime.strptime(data, "%d/%m/%Y")
    return data
```

```python
pedido = conversao_data(data_pedido)
entrega = conversao_data(data_entrega)
print(f"Pedido: {pedido}\nEntrega: {entrega}")
```

```python
dias = entrega.day - pedido.day
print(dias)
```

```python
def conversao_data_texto(data):
    data = data.strftime("%d/%m/%Y")
    return data
```

```python
data_pedido_convertido = conversao_data_texto(pedido)
data_entrega_convertido = conversao_data_texto(entrega)
print(f"Pedido realizado em: {data_pedido_convertido}\nEntrega realizada em: {data_entrega_convertido}")
print(f"Tempo de entrega: {dias} dias")
if dias > 5:
    print(f"A entrega ocorreu em {dias} dias, infelizmente ficou acima de 5 dias, que decepção, viu!")
else:
    print(f"A entrega ocorreu em {dias} dias, portanto, ficou dentro do prazo")
```

### **Prática: Verificação de Produtividade dos funcionários**

```python
vendedores = [
    {"nome": "Ana", "vendas": 12500},
    {"nome": "Carlos", "vendas": 9800},
    {"nome": "Marina", "vendas": 15700},
    {"nome": "Rafael", "vendas": 11300}
]
```

```python
print("Verificação de Produtividade dos Funcionários!")
for numero, vendedor in enumerate(vendedores, start=1):
    print(f"{numero} - {vendedor['nome']}: R${vendedor['vendas']}")
    if vendedor['vendas'] >= 10000:
        print("Situação: Meta atingida!\n")
    else:
        print("Situação: Meta não atingida!\n")
```

### **Prática - Registro de participantes**

```python
def registrar_treinamento(tema, *participantes):
    print("Tema:", tema)
    for numero, participante in enumerate(participantes, start=1):
        print(f"{numero} - {participante}")
```

```python
registrar_treinamento(
    "Segurança da Informação",
    "José", 
    "Maria",
    "Pedro"
)
```

```python
registrar_treinamento(
    "Python",
    "Paulo"
)
```
