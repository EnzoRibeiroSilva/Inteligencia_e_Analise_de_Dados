Problema 1

```python
estoque = [18, 5, 0, 32, 4, 7, 0, 14, 3]
produto_esgotado = 0
produto_critico = 0

for i in range(len(estoque)):
    if estoque[i] == 0:
        print(f"O produto {i+1} está esgotado!")
        produto_esgotado += 1

    elif estoque[i] > 0 and estoque[i] <= 5:
        print(f"O produto {i+1} está com o estoque crítico!")
        produto_critico += 1

    elif estoque[i] > 5:
        print(f"O produto {i+1} está com o estoque disponível!")

print(f"\nQuantidade de produtos esgotados: {produto_esgotado}")
print(f"Quantidade de produtos críticos: {produto_critico}")

percentual = ((produto_esgotado + produto_critico) / len(estoque)) * 100

print(f"Percentual dos produtos que precisam de reposição: {round(percentual, 2)}%")
```

Problema 3

```python
atendimentos = [
    [120, 135, 128, 141],
    [98, 105, 110, 103],
    [150, 145, 160, 155],
    [116, 161, 108, 121]
]

soma = 0
maior = 0
lista_soma = []
abaixo = 0
pos = 0

print("Desempenho das Unidades!")

for i in range(len(atendimentos)):
    soma = 0

    for j in range(len(atendimentos)):
        soma += atendimentos[i][j]   

        if maior < atendimentos[i][j]:
            maior, atendimentos[i][j] = atendimentos[i][j], maior

        if atendimentos[i][j] < 110:
            abaixo += 1

    lista_soma.append(soma)

    maior_unidade = max(lista_soma)
    if lista_soma[i] == maior_unidade:
        pos = i
    
    print(f"Unidade {i+1} - Total de atendimentos: {soma} unidades")

print(f"\nA unidade que realizou mais atendimentos no período foi: Unidade {pos+1}")
print(f"Quantidade de registros abaixo de 110 atendimentos: {abaixo} registros")
print(f"A maior quantidade de registros realizada em 1 semana foi: {maior} atendimentos")
```

```python
produtos = {
    "Produto": "Notebook",
    "Valor": 4200,
    "Estoque": 8
}

for chave in produtos.keys():
    print(f"Chave - {chave}")

print()

for valor in produtos.values():
    print(f"Valor - {valor}")

print()

for chave, valor in produtos.items():
    print(f"{chave}: {valor}")
```

Prática: For + dicionário + if

```python
produtos = [
    {
        "Produto": "Notebook",
        "Preço": 4200,
        "Estoque": 8
     },

     {
        "Produto": "Mouse",
        "Preço": 120,
        "Estoque": 25
    },

    {
        "Produto": "Monitor",
        "Preço": 1350,
        "Estoque": 4
     },

     {
        "Produto": "Teclado",
        "Preço": 280,
        "Estoque": 15
    },

    {
        "Produto": "Headset",
        "Preço": 450,
        "Estoque": 3
    },
]

cont = 0

print("Produtos cadastrados!")
for dicionarios in produtos:
    for chave, valor in dicionarios.items():
        print(f"{chave}: {valor}")
    print()

print("Produtos com estoque inferior a 10 unidades!")
for dicionarios in produtos:
    for chave, valor in dicionarios.items():
        if chave == "Estoque" and dicionarios[chave] < 10:
            cont += 1
            print(f"Produto: {dicionarios["Produto"]} - Estoque: {dicionarios[chave]} unidades")

print(f"\nQuantidade de produtos que precisam de reposição: {cont} produtos")
    
```
