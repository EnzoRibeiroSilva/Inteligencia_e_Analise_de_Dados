Problema 3

```python
producoes = [485, 510, 472, 498, 455, 520]
cont = 0

quant_turnos = len(producoes)
producao_total = sum(producoes)
producao_media = producao_total / quant_turnos
maior_producao = max(producoes)
menor_producao = min(producoes)

print("Monitoramento dos Turnos de Produção")
print(f"Quantidade de turnos: {quant_turnos} turnos")
print(f"Produção Total: {producao_total} produções")
print(f"Produção Média: {producao_media:.0f} produções")
print(f"Maior Produção: {maior_producao} produções")
print(f"Menor Produção: {menor_producao} produções")

if producao_media < 480:
    print("Desempenho da produção: Abaixo da meta")
elif producao_media >= 480 and producao_media < 500:
    print("Desempenho da produção: Atenção!")
else:
    print("Desempenho da produção: Meta atingida!")

for i in producoes:
    if i < 460:
        cont += 1
if cont >= 1:
    print(f"É necessário emitir um alerta operacional! {cont} turno(s) foram identificados com a produção inferior a 460 unidades!")
else:
    print("Não é necessário emitir um alerta operacional!")
```

```python
janeiro = [85000, 46000, 18000]
fevereiro = [92000, 51000, 19000]
marco = [78000, 48000, 20000]
abril = [105000, 54000, 21000]
maio = [88000, 52000, 22000]
imo = []
classi = []

resultado_janeiro = janeiro[0] - janeiro[1] - janeiro[2]
resultado_fevereiro = fevereiro[0] - fevereiro[1] - fevereiro[2]
resultado_marco = marco[0] - marco[1] - marco[2]
resultado_abril = abril[0] - abril[1] - abril[2]
resultado_maio = maio[0] - maio[1] - maio[2]

imo.append((resultado_janeiro / janeiro[0]) * 100)
imo.append((resultado_fevereiro / fevereiro[0]) * 100)
imo.append((resultado_marco / marco[0]) * 100)
imo.append((resultado_abril / abril[0]) * 100)
imo.append((resultado_maio / maio[0]) * 100)

if imo[0] < 0:
    classi.append('Prejuízo')
elif imo[0] == 0:
    classi.append('Ponto de equilíbrio')
elif imo[0] > 0 and imo[0] < 8:
    classi.append('Atenção')
elif imo[0] >= 8 and imo[0] < 15:
    classi.append('Regular')
elif imo[0] >= 15 and imo[0] < 25:
    classi.append('Bom')
else:
    classi.append('Excelente')

if imo[1] < 0:
    classi.append('Prejuízo')
elif imo[1] == 0:
    classi.append('Ponto de equilíbrio')
elif imo[1] > 0 and imo[1] < 8:
    classi.append('Atenção')
elif imo[1] >= 8 and imo[1] < 15:
    classi.append('Regular')
elif imo[1] >= 15 and imo[1] < 25:
    classi.append('Bom')
else:
    classi.append('Excelente')

if imo[2] < 0:
    classi.append('Prejuízo')
elif imo[2] == 0:
    classi.append('Ponto de equilíbrio')
elif imo[2] > 0 and imo[2] < 8:
    classi.append('Atenção')
elif imo[2] >= 8 and imo[2] < 15:
    classi.append('Regular')
elif imo[2] >= 15 and imo[2] < 25:
    classi.append('Bom')
else:
    classi.append('Excelente')

if imo[3] < 0:
    classi.append('Prejuízo')
elif imo[3] == 0:
    classi.append('Ponto de equilíbrio')
elif imo[3] > 0 and imo[3] < 8:
    classi.append('Atenção')
elif imo[3] >= 8 and imo[3] < 15:
    classi.append('Regular')
elif imo[3] >= 15 and imo[3] < 25:
    classi.append('Bom')
else:
    classi.append('Excelente')

if imo[4] < 0:
    classi.append('Prejuízo')
elif imo[4] == 0:
    classi.append('Ponto de equilíbrio')
elif imo[4] > 0 and imo[4] < 8:
    classi.append('Atenção')
elif imo[4] >= 8 and imo[4] < 15:
    classi.append('Regular')
elif imo[4] >= 15 and imo[4] < 25:
    classi.append('Bom')
else:
    classi.append('Excelente')

print("Relatório de Rentabilidade")
print(f"Resultado financeiro de Janeiro: R${resultado_janeiro}.\nIMO de Janeiro: {imo[0]:.2f}.\nClassificação: {classi[0]}\n")
print(f"Resultado financeiro de Fevereiro: R${resultado_fevereiro}.\nIMO de Fevereiro: {imo[1]:.2f}.\nClassificação: {classi[1]}\n")
print(f"Resultado financeiro de Março: R${resultado_marco}.\nIMO de Março: {imo[2]:.2f}.\nClassificação: {classi[2]}\n")
print(f"Resultado financeiro de Abril: R${resultado_abril}.\nIMO de Abril: {imo[3]:.2f}.\nClassificação: {classi[3]}\n")
print(f"Resultado financeiro de Maio: R${resultado_maio}.\nIMO de Maio: {imo[4]:.2f}.\nClassificação: {classi[4]}\n")
print(f"Maior IMO: {max(imo)}. Correspondente ao mês de Abril!")
print(f"Menor IMO: {min(imo)}. Correspondente ao mês de Março!")

```
