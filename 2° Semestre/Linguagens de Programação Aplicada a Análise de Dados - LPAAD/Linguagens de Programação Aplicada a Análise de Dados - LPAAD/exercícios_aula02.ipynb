Contador + Somatório

```python
notas = [8.5, 4.0, 7.5, 5.0, 9.0]
cont = 0
soma = 0

for nota in notas:
    if nota >= 6:
        cont += 1
        soma += nota

media = soma / cont
print(f"Temos {cont} notas maiores ou igual a 6. \nA média delas é: {round(media)}")
```

Prática - Uma empresa registrou o tempo em minutos, necessário para realizar seis entregas

```python
tempos = [32, 48, 27, 55, 41, 29]
cont_atrasada = 0
soma = 0
soma_falhas = 0

print("Relatório do Tempo das Entregas!\n")
for i in range(len(tempos)):
    print(f"Entrega {i+1}\nTempo de entrega: {tempos[i]} minutos")
    soma += tempos[i]

    if tempos[i] > 40:
        print("Condição da entrega: Atrasada\n")
        cont_atrasada += 1
        soma_falhas += tempos[i]
        
    else:
        print("Condição da entrega: Dentro do Prazo\n")

media = soma / len(tempos)
porcentagem = (cont_atrasada * 100) / len(tempos)

print(f"Tempo total das entregas: {soma} minutos!")
print(f"Tempo médio das entregas: {round(media,2)} minutos!")
print(f"Quantidade de entregas atrasadas: {cont_atrasada} entregas!")
print(f"Porcentagem Correspondente as entregas atrasadas: {round(porcentagem)}%")
```

Prática: Controle de Produção

```python
producoes = []
soma = 0
cont_meta = 0

print("Registro da Produção de Peças!")
quant = int(input("Digite a quantidade de peças produzidas em cada turno (Entre 300 e 800): "))
while quant >= 300 and quant <= 800:
    producoes.append(quant)
    soma += quant
    quant = int(input("Digite a quantidade de peças produzidas em cada turno (Entre 300 e 800): "))

for i in range(len(producoes)):
    if producoes[i] >= 500:
        print(f"Turno {i+1} - {producoes[i]} peças: Meta atingida!")
        cont_meta += 1
    elif producoes[i] >= 400 and producoes[i] < 500:
        print(f"Turno {i+1} - {producoes[i]} peças: Abaixo da Meta!")
    else:
        print(f"Turno {i+1} - {producoes[i]} peças: Crítico!")

turnos = len(producoes)
media = soma / turnos

percentual = (cont_meta * 100) / turnos

print("\nRelatório de Produção")
print(f"Produção Total: {soma} peças")
print(f"Quantidade de turnos: {turnos} turnos")
print(f"Produção Média: {round(media,2)} peças")
print(f"Quantidade de turnos que atingiram a meta: {cont_meta} turnos")
print(f"Percentual de turnos que atingiram a meta: {round(percentual,2)}%")
```
