# **Aula de Programação a Orientação de Objetos**

- Exemplo: As funções utilizadas no improt de uma biblioteca representa métodos de uma classe/objeto.

Anotações:
- Entidade - Objeto onde farei o molde dele (Exemplo: Música, jogo, filme);
- Atributo - São as características dessa garrafa, desse objeto específico;
- No momento em que você está definindo um objeto, é importante definir os atributos que são apenas necessários;
- Método - Toda vez que você faz uma ação, você está fazendo um método;
- Objeto e instância são as mesmas coisas, porém semanticamente diferentes pela separação de um clique;
- Objeto: O ato de alocar a estrutura de um objeto em uma variável (Informalmente falando) - Todo objeto é independente, mesmo tendo mesmos atributos;
- Instância, é um objeto criado a partir de uma classe, o ato de criar o objeto é fazer uma instÂncia, o momento em que você termina esse processo, em que você alocou lá, você fez um objeto. Toda vez que você leva essas informações da classe para a variável, você está criando o objeto;
- Instância faz a transformação do abstrato para o real, classe para objeto.

- Métodos (Operações): São ações de um objeto, portanto não precisa ser definido na função

# ****

```python
class Musica:
    def __init__(self, nome, cantor, genero, duracao, ano):
        self.nome = nome
        self.cantor = cantor
        self.genero = genero
        self.duracao = duracao
        self.ano = ano

        # Atributos de controle
        self.reproducoes = 0
        self.tocando = False
    
    # Padronize os nomes das funções
    # Evite colocar o nome da função com o mesmo nome do atributo
    def exibicao(self):
        print(f"Nome da Música: {self.nome}\nCantor: {self.cantor}\nGênero: {self.genero}\nDuração: {self.duracao} segundos\nAno de Lançamento: {self.ano}\nNúmero de Reproduções: {self.reproducoes}")
    
    def reproduzir(self):
        self.reproducoes += 1
        print(f"Reproduzindo: {self.nome}")
    
    def verificar_reproducao(self):
        if self.tocando == False:
             self.tocando = True
             self.reproducoes += 1
             print(f"Reproduzindo: {self.nome}")
        else:
            print("Essa música já sendo reproduzida!")

    def pausar(self):
        if self.tocando == True:
            self.tocando == False
            print("Música Pausada!")

        else:
            print("Sem música em reprodução")

```

```python
m1 = Musica("Die With a Smile", "Bruno Mars", "Pop", 130, 2024)
```

```python
m1.reproduzir()
```

```python
print("Informações Detalhadas")
m1.exibicao()
```

```python
m1.pausar()
m1.reproduzir()
print(m1.reproducoes)
```
