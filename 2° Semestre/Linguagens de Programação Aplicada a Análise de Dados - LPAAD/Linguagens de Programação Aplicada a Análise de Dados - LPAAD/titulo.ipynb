# **Prática 1 - Operação Bancária**

```python
while True:
    while True:
        try:
            saque = float(input("Digite o valor desejado para o saque (entre 100 e 5000): "))
            break
        except ValueError:
            print("Formato não aceito, insira o valor desejado!") 
    limite = 8000
    if saque < 100 or saque > 5000:
        print("Valor de saque não permitido, tente novamente no intervalo indicado!")
    else:
        new_limite = limite - saque
        if new_limite < 3500:
            print("Valor de limite inferior ao piso estabelecido de acordo com o valor alto de saque, tente novamente!")
        else:
            while True:
                try:
                    quant_parcelas = int(input("Defina o número de parcelas desejadas de 1 a 6:"))
                    break
                except ValueError:
                    print("Formato não aceito, insira o valor desejado!")  
            if quant_parcelas < 1 or quant_parcelas > 6:
                print("Número de parcelas inválido, coloque no intervalo indicado!")
            else:
                parcela = saque / quant_parcelas
                if parcela < 50:
                    print(f"Valor mínimo de parcela abaixo de 50 reais, valor da parcela calculado: R${parcela:.2f}\nTente novamente!")
                else:
                    print(f"\nValor de saque: R${saque:.2f}\nNovo limite: R${new_limite:.2f}\nQuantidade de Parcelas: {quant_parcelas}\nValor da parcela: R${parcela:.2f}")
                    print("\nOperação finalizada, obrigado pela preferência!")
                    break
```

# **Prática 1.2**

```python
while True:
    while True:
        try:
            saque = float(input("Digite o valor desejado para o saque (entre 100 e 5000): "))
            break
        except ValueError:
            print("Formato não aceito, insira o valor desejado!") 
    limite = 8000
    if saque < 100 or saque > 5000:
        print("Valor de saque não permitido, tente novamente no intervalo indicado!")
    else:
        while True:
            try:
                quant_parcelas = int(input("Defina o número de parcelas desejadas de 1 a 6:"))
                break
            except ValueError:
                print("Formato não aceito, insira o valor desejado!")  
        if quant_parcelas < 1 or quant_parcelas > 6:
            print("Número de parcelas inválido, coloque no intervalo indicado!")
        else:
            tipo_juros = int(input("Digite o tipos de juros desejados [1 - Juros Simples / 2 - Juros Compostos]:"))
            taxa = 0.02

            if tipo_juros == 1:
                juros = saque * taxa * quant_parcelas
                parcela = (saque / quant_parcelas) + (juros / quant_parcelas)

            elif tipo_juros == 2:
                juros = saque * (1 + taxa) ** quant_parcelas
                parcela = (saque / quant_parcelas) + (juros / quant_parcelas)
            
            new_limite = limite - (parcela * quant_parcelas)
            if new_limite < 3500:
                print("Valor de limite inferior ao piso estabelecido de acordo com o valor alto de saque, tente novamente!")
            else:
                if parcela < 50:
                    print(f"Valor mínimo de parcela abaixo de 50 reais, valor da parcela calculado: R${parcela:.2f}\nTente novamente!")
                else:
                    print(f"\nValor de saque: R${saque:.2f}\nNovo limite: R${new_limite:.2f}\nQuantidade de Parcelas: {quant_parcelas}\nValor da parcela: R${parcela:.2f}\nTotal de juros: R${juros:.2f}\nJuros em cada parcela: R${juros / quant_parcelas:.2f}")
                    print("\nOperação finalizada, obrigado pela preferência!")
                    break
    
```
