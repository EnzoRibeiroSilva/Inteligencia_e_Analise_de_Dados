Importação

```python
from cryptography.fernet import Fernet
```

Gerar uma chave secreta

```python
chave = Fernet.generate_key()
print(chave)
```

Criptografar

```python
cifra = Fernet(chave)
dados = """
Cliente: Empresa ABC
CNPJ: 12.345.678/0001-90
Contrato: R$ 850.00
"""

dados_cifrados = cifra.encrypt(dados.encode())
print(dados_cifrados)
```

Descriptografar

```python
dados_decifrados = cifra.decrypt(dados_cifrados)
print(dados_decifrados.decode('utf-8'))
```
